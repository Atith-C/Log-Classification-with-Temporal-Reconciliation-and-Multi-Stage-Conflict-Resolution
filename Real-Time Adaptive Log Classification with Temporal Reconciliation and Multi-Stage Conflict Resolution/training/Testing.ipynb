# 1. Precision of Regex

```python
import pandas as pd
from processor_regex import classify_with_regex
from sklearn.metrics import precision_score

# Load dataset
df = pd.read_excel("dataset/test.xlsx")
print(df.shape)

# True labels
y_true = df["true_label"].tolist()

# Regex predictions
y_pred = [classify_with_regex(log) for log in df["log_message"].tolist()]

# Compute precision (ignore None predictions)
valid_idx = [i for i, p in enumerate(y_pred) if p is not None]
precision = precision_score([y_true[i] for i in valid_idx],
                            [y_pred[i] for i in valid_idx],
                            average="macro")

print(f"Regex Precision: {precision:.2f}")
```

# 2. Accuracy of ML

```python
from processor_bert import classify_with_bert
from sklearn.metrics import accuracy_score

# True labels
y_true = df["true_label"].tolist()

# ML predictions
y_pred = [classify_with_bert(log) for log in df["log_message"].tolist()]

# Accuracy
acc = accuracy_score(y_true, y_pred)
print(f"ML Accuracy: {acc:.2f}")
```

# 3. Hybrid Improvement

```python
from classify import classify

# True labels
y_true = df["true_label"].tolist()

# Hybrid predictions
y_pred = classify(list(zip(df["source"], df["log_message"])))

# Accuracy
acc = accuracy_score(y_true, y_pred)
print(f"Hybrid Accuracy: {acc:.2f}")
```

# 4. Overall Coverage

```python
import numpy as np

# Coverage = % of logs with a non-null prediction
coverage = np.mean([p is not None for p in y_pred]) * 100
print(f"Hybrid Coverage: {coverage:.2f}%")
```
